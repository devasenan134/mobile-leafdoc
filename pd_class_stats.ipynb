```python
import argparse, json
from pathlib import Path
from collections import Counter
```

```python
EXTS={".jpg",".jpeg",".png",".bmp",".tif",".tiff",".webp"}
```

```python
class_map = "classmap.json"
pd_train_dir = "../PlantDoc-Dataset/train/"
```

```python
cm=json.load(open(class_map))
u2i={u:i for i,u in enumerate(cm["unified_classes"])}
pd2u=cm["plantdoc_to_unified"]

cnt=Counter()
for d in sorted([p for p in Path(pd_train_dir).iterdir() if p.is_dir()]):
    if d.name not in pd2u: continue
    u=pd2u[d.name]
    if u not in u2i: continue
    for im in d.rglob("*"):
        if im.suffix.lower() in EXTS: cnt[u]+=1

print("PD-train per-class counts:")
for u in cm["unified_classes"]:
    if u in cnt: print(f"{u:35s} {cnt[u]}")
print("\nTotal:", sum(cnt.values()))
```
